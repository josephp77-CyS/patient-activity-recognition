```python
import zipfile
import os

# 1. Buat folder khusus penampung dataset jika belum ada
folder_tujuan = "UCI_HAR_Dataset"
os.makedirs(folder_tujuan, exist_ok=True)

# 2. Ekstrak file zip ke dalam folder tersebut
zip_filename = 'UCI_HAR_Dataset.zip'

if os.path.exists(zip_filename):
    print(f"Mengekstrak {zip_filename} ke dalam folder '{folder_tujuan}'...")
    with zipfile.ZipFile(zip_filename, 'r') as zip_ref:
        zip_ref.extractall(folder_tujuan)
    print("Ekstraksi berhasil!")
else:
    print("File ZIP tidak ditemukan!")

# 3. Cek kembali isi folder tujuannya
print("Isi folder:", os.listdir(folder_tujuan))
```
